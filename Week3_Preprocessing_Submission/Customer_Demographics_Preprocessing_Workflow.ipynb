```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
print("Age data type:", df["Age"].dtype)
```

```python
print(df["Age"].value_counts().head(30))
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
print("Age data type:", df["Age"].dtype)
print("Missing values:", df["Age"].isna().sum())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
```

```python
print(df["Age"].value_counts().sort_index().head(20))
```

```python
print(df["Age"].value_counts().sort_index().tail(20))
```

```python
print("\nNumber of ages below 0:", (df["Age"] < 0).sum())
print("Number of ages above 100:", (df["Age"] > 100).sum())
```

```python
print("Ages below 0:")
print(df.loc[df["Age"] < 0, "Age"].value_counts())

print("\nAges above 100:")
print(df.loc[df["Age"] > 100, "Age"].value_counts())
```

```python
original_df = pd.read_csv("/content/drive/MyDrive/MMLA Datasets/FinMark CSV/customer_demographics_contaminated.csv")

numeric_age = pd.to_numeric(original_df["Age"], errors="coerce")

print("Non-numeric Age values:")
print(original_df.loc[
    numeric_age.isna() & original_df["Age"].notna(),
    "Age"
].value_counts())
```

```python
df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = pd.NA
```

```python
print("Missing/invalid Age values:", df["Age"].isna().sum())
print("Minimum valid Age:", df["Age"].min())
print("Maximum valid Age:", df["Age"].max())
```

```python
print("Mean Age:", df["Age"].mean())
print("Median Age:", df["Age"].median())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
print("Missing Age values:", df["Age"].isna().sum())
print("Age data type:", df["Age"].dtype)
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
```

```python
print("IncomeLevel data type:", df["IncomeLevel"].dtype)

print("\nIncomeLevel categories:")
print(df["IncomeLevel"].value_counts(dropna=False))
```

```python
df["IncomeLevel"] = df["IncomeLevel"].fillna("Undetermined")
```

```python
print("Missing IncomeLevel values:", df["IncomeLevel"].isna().sum())

print("\nIncomeLevel categories:")
print(df["IncomeLevel"].value_counts())
```

```python
print("SignupDate data type:", df["SignupDate"].dtype)
```

```python
print(df["SignupDate"].head(20))
```

```python
print(df["SignupDate"].tail(20))
```

```python
raw_dates = df["SignupDate"].astype("string").str.strip()

parsed_dates = pd.to_datetime(
    raw_dates,
    format="%Y-%m-%d",
    errors="coerce"
)

mask = parsed_dates.isna()

parsed_dates.loc[mask] = pd.to_datetime(
    raw_dates.loc[mask],
    format="%d/%m/%Y",
    errors="coerce"
)

df["SignupDate"] = parsed_dates
```

```python
print("SignupDate data type:", df["SignupDate"].dtype)
print("Invalid SignupDate values:", df["SignupDate"].isna().sum())
```

```python
print("FINAL DATASET CHECK")

print("\nShape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
Customer_Demographics_Cleaned_Dataset = "Customer_Demographics_Cleaned_Dataset.csv"

df.to_csv(Customer_Demographics_Cleaned_Dataset, index=False)

print("Save as:", Customer_Demographics_Cleaned_Dataset)
```
