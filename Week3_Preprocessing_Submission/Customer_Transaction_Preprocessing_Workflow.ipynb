```python
df = df.copy()
```

```python
df["Amount"] = df["Amount"].replace("Free", 0)
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
```

```python
df.loc[df["Amount"] == -100, "Amount"] = pd.NA
```

```python
print("Amount data type:", df["Amount"].dtype)
print("Invalid values:", df["Amount"].isna().sum())
print("Negative values:", (df["Amount"] < 0).sum())
```

```python
median_amount = df["Amount"].median()

print(median_amount)
```

```python
df["Amount"] = df["Amount"].fillna(median_amount)
```

```python
print("Missing:", df["Amount"].isna().sum())
print("Minimum Amount value:", df["Amount"].min())
print("Median:", df["Amount"].median())
```

```python
df["ProductCategory"] = df["ProductCategory"].fillna("Undetermined")
```

```python
print("Missing values:", df["ProductCategory"].isna().sum())
print("\nProductCategory values:")
print(df["ProductCategory"].value_counts())
```

```python
df["TransactionDate"] = pd.to_datetime(
    df["TransactionDate"],
    format="%Y-%m-%d",
    errors="coerce"
)
```

```python
print( df["TransactionDate"].dtype)
print("Invalid:", df["TransactionDate"].isna().sum())
```

```python
original_df = pd.read_csv("/content/drive/MyDrive/MMLA Datasets/FinMark CSV/customer_transactions_contaminated.csv")

invalid_dates = original_df[
    pd.to_datetime(
        original_df["TransactionDate"],
        format="%Y-%m-%d",
        errors="coerce"
    ).isna()
]

print("Invalid values:", len(invalid_dates))
print(invalid_dates["TransactionDate"].value_counts())
```

```python
df["TransactionDate"] = original_df["TransactionDate"].astype("string").str.strip()

parsed_dates = pd.to_datetime(
    df["TransactionDate"],
    format="%Y-%m-%d",
    errors="coerce"
)

mask = parsed_dates.isna()

parsed_dates.loc[mask] = pd.to_datetime(
    df.loc[mask, "TransactionDate"],
    format="%d/%m/%Y",
    errors="coerce"
)

df["TransactionDate"] = parsed_dates
```

```python
print("TransactionDate data type:", df["TransactionDate"].dtype)
print("Invalid date values:", df["TransactionDate"].isna().sum())
```

```python
print("FINAL DATASET CHECK")

print("\nShape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
Customer_Transaction_Cleaned_Dataset = "Customer_Transaction_Cleaned_Dataset.csv"

df.to_csv(Customer_Transaction_Cleaned_Dataset, index=False)

print("Save as:", Customer_Transaction_Cleaned_Dataset)
```
